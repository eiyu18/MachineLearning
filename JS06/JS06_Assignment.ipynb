```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

df = pd.read_csv("/content/sample_data/voice.csv")
X = df.drop(columns="label")
y = df["label"].map({"male": 1, "female": 0})

rows = []
for test_size, name in [(0.30, "70:30"), (0.20, "80:20")]:
    X_tr, X_te, y_tr, y_te = train_test_split(
        X, y, test_size=test_size, random_state=42, stratify=y
    )
    sc = StandardScaler().fit(X_tr)
    X_tr_s, X_te_s = sc.transform(X_tr), sc.transform(X_te)
    for kernel in ["linear", "poly", "rbf"]:
        m = SVC(kernel=kernel).fit(X_tr_s, y_tr)
        rows.append((name, kernel, accuracy_score(y_te, m.predict(X_te_s))))

res = pd.DataFrame(rows, columns=["split", "kernel", "accuracy"])
print(res.pivot(index="kernel", columns="split", values="accuracy").round(4))
```
